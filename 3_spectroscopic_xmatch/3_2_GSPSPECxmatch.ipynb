```python
import pandas as pd
from compute_luminosity import compute_lum
```

# **Generate GSP-Spec Input Catalogue**

Read-in asteroseismic and GSP-Spec data.

```python
old_data = pd.read_csv("./gaia_asteroseismic.csv")
old_data = old_data.dropna(subset=["GSPSPEC-teff", "GSPSPEC-e_teff", "GSPSPEC-logg", "GSPSPEC-e_logg", "GSPSPEC-fe_h", "GSPSPEC-e_fe_h"])
```

Tidy columns in dataframe such that only the required remain.

```python
new_data = old_data.drop(columns=["GAIA-teff", "GAIA-e_teff", "GAIA-logg", "GAIA-e_logg", "GAIA-fe_h", "GAIA-e_fe_h", "GAIA-plx", "GAIA-e_plx", "gdr3_id", "BJ-rgeo", "BJ-rphoto", "BJ-flag", "BJ-e_rgeo", "BJ-e_rphoto"])
new_data.columns = new_data.columns.str.replace("GSPSPEC-", "", regex=False)
new_data.columns = new_data.columns.str.replace("GAIA-", "", regex=False)
```

Compute luminosity values.

```python
lum_array, e_lum_array = compute_lum("GSPSPEC", new_data["a0"], new_data["teff"], new_data["gmag"], new_data["e_gmag"], new_data["dist"], new_data["e_dist"], new_data["logg"], new_data["fe_h"],)
```

Write to file.

```python
new_data["lum"] = lum_array
new_data["e_lum"] = e_lum_array
new_data.to_csv("./GSPSPEC_input_catalogue.csv", index=False)
```
