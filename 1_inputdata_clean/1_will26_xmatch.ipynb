```python
import pandas as pd
import numpy as np
from io import StringIO
```

## **Willett+ 2026**

### Data Description

### Read-In Data

Read-in data.

```python
pre_radec = pd.read_csv("full_catalogues/Will2026.csv")
```

```python
pd.DataFrame(pre_radec["K2_ID"][:5000], dtype="int").to_csv("id_lists/w26_epic_1.csv", index=False)
pd.DataFrame(pre_radec["K2_ID"][5000:], dtype="int").to_csv("id_lists/w26_epic_2.csv", index=False)
```

Use K2 Search and Retrieve to find RA and Dec from K2 IDs. Read-in results.

```python
rows = []

with open("partial_catalogues/w26_radec_1.txt") as f:
    for i, line in enumerate(f):
        if i % 2 == 1:  # every second line
            rows.append(line.strip())

with open("partial_catalogues/w26_radec_2.txt") as f:
    for i, line in enumerate(f):
        if i % 2 == 1:
            rows.append(line.strip())

radec = pd.read_csv(
    StringIO("\n".join(rows)),
    header=None,
    names=["K2_ID", "RA", "DEC"]
)

#radec = radec[radec["K2_ID"] != "target not resolved"]

radec["K2_ID"] = radec["K2_ID"].astype("Int64")
```

Merge RA and Dec values.

```python
post_radec = pd.merge(pre_radec, radec, on="K2_ID", how="left")
```

Write catalogue to `.csv` file for CDS xmatching.

```python
post_radec.to_csv("partial_catalogues/w26_precds.csv", index=False)
```

### CDS X-Matching

Read-in xmatch.

```python
array_full = pd.read_csv("full_catalogues/Will2026_cdsxmatch.csv")
```

Remove flagged entries.

```python
array_full = array_full[
    (array_full["warn_low_numax"] == False) &
    (array_full["warn_high_Dnu"] == False) &
    (array_full["warn_high_RUWE"] == False)
]
print(f"Cleaned data length: {len(array_full):.0f}")
```

Curate simplfied array.

```python
w26 = pd.DataFrame({
    "gaiadr3_id": array_full["Source"],
    "ra": array_full["RAdeg"],
    "dec": array_full["DEdeg"],
    "numax": array_full["numax"],
    "e_numax": array_full["numax_err"],
    "dnu": array_full["Dnu"],
    "e_dnu": array_full["Dnu_err"],
    "ev_state": np.full(len(array_full), np.nan),
    "teff": array_full["Teff"],
    "e_teff": (array_full["B_Teff"] - array_full["b_Teff"]) / 2,
    "logg": array_full["logg"],
    "e_logg": (array_full["B_logg"] - array_full["b_logg"]) / 2,
    "fe_h": array_full["[Fe/H]"],
    "e_fe_h": (array_full["B_[Fe/H]"] - array_full["b_[Fe/H]"]) / 2,
    "GSPSPEC-fe_h": array_full["[Fe/H]-S"],
    "GSPSPEC-e_fe_h": (array_full["B_[Fe/H]-S"] - array_full["b_[Fe/H]-S"]) / 2,
    "GSPSPEC-si_fe": array_full["[Si/Fe]-S"],
    "GSPSPEC-e_si_fe": (array_full["B_[Si/Fe]-S"] - array_full["b_[Si/Fe]-S"]) / 2,
    "GSPSPEC-ca_fe": array_full["[Ca/Fe]-S"],
    "GSPSPEC-e_ca_fe": (array_full["B_[Ca/Fe]-S"] - array_full["b_[Ca/Fe]-S"]) / 2,
    "GSPSPEC-ti_fe": array_full["[Ti/Fe]-S"],
    "GSPSPEC-e_ti_fe": (array_full["B_[Ti/Fe]-S"] - array_full["b_[Ti/Fe]-S"]) / 2,
    "GSPSPEC-mg_fe": array_full["[Mg/Fe]-S"],
    "GSPSPEC-e_mg_fe": (array_full["B_[Mg/Fe]-S"] - array_full["b_[Mg/Fe]-S"]) / 2,
    "GSPSPEC-nd_fe": array_full["[Nd/Fe]-S"],
    "GSPSPEC-e_nd_fe": (array_full["B_[Nd/Fe]-S"] - array_full["b_[Nd/Fe]-S"]) / 2,
    "GSPSPEC-feii_m": array_full["[FeII/M]-S"],
    "GSPSPEC-e_feii_m": (array_full["B_[FeII/M]-S"] - array_full["b_[FeII/M]-S"]) / 2,
    "GSPSPEC-s_fe": array_full["[S/Fe]-S"],
    "GSPSPEC-e_s_fe": (array_full["B_[S/Fe]-S"] - array_full["b_[S/Fe]-S"]) / 2,
    "GSPSPEC-zr_fe": array_full["[Zr/Fe]-S"],
    "GSPSPEC-e_zr_fe": (array_full["B_[Zr/Fe]-S"] - array_full["b_[Zr/Fe]-S"]) / 2,
    "GSPSPEC-n_fe": array_full["[N/Fe]-S"],
    "GSPSPEC-e_n_fe": (array_full["B_[N/Fe]-S"] - array_full["b_[N/Fe]-S"]) / 2,
    "GSPSPEC-cr_fe": array_full["[Cr/Fe]-S"],
    "GSPSPEC-e_cr_fe": (array_full["B_[Cr/Fe]-S"] - array_full["b_[Cr/Fe]-S"]) / 2,
    "GSPSPEC-ce_fe": array_full["[Ce/Fe]-S"],
    "GSPSPEC-e_ce_fe": (array_full["B_[Ce/Fe]-S"] - array_full["b_[Ce/Fe]-S"]) / 2,
    "GSPSPEC-ni_fe": array_full["[Ni/Fe]-S"],
    "GSPSPEC-e_ni_fe": (array_full["B_[Ni/Fe]-S"] - array_full["b_[Ni/Fe]-S"]) / 2
})
```

Write to `.csv` file.
